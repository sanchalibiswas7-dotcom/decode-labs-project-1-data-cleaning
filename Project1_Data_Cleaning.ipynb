# Data Analytics Internship Project

## Project 1: Data Cleaning & Preparation

```python
import pandas as pd
```

```python
df = pd.read_excel("/content/Dataset for Data Analytics.xlsx")

df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df["CouponCode"] = df["CouponCode"].fillna("No Coupon")
```

```python
print("Duplicate rows:", df.duplicated().sum())
df = df.drop_duplicates()
```

```python
text_columns = [
    "Product",
    "PaymentMethod",
    "OrderStatus",
    "ReferralSource"
]

for col in text_columns:
    df[col] = df[col].str.strip().str.title()
```

```python
df["Date"] = pd.to_datetime(df["Date"])
df["Date"] = df["Date"].dt.strftime("%Y-%m-%d")
```

```python
df["CalculatedTotal"] = df["Quantity"] * df["UnitPrice"]

mismatch = df[df["CalculatedTotal"] != df["TotalPrice"]]

print(mismatch)
```

```python
df["CalculatedTotal"] = (df["Quantity"] * df["UnitPrice"]).round(2)

mismatch = df[df["CalculatedTotal"] != df["TotalPrice"]]

print(mismatch)
```

```python
mismatch = df[abs(df["CalculatedTotal"] - df["TotalPrice"]) > 0.01]

print(mismatch)
```

```python
df.to_excel("Cleaned_Ecommerce_Dataset.xlsx", index=False)

print("Dataset cleaned successfully!")
```
